```python
# extraction of digits using loops

n = 5873

while n>0:
    last_digits = n%10
    print(last_digits)
    n = n//10
    
```

```python
n = 5873
count = 0

while n > 0:
    count += 1
    n = n // 10

print(count)
```

```python
# check palindrome

n = 1234
result = 0

while n>0:
    id = n%10
    result = (result*10)+id
    n = n//10

```
