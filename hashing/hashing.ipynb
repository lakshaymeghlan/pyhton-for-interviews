```python
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]

# Hash array (0 to 10)
hash_list = [0] * 11

# Step 1: Build frequency hash from n
for x in n:
    if x < len(hash_list):
        hash_list[x] += 1

# Step 2: Answer queries from m
for q in m:
    if q < 0 or q >= len(hash_list):
        print(0)
    else:
        print(hash_list[q])
```
