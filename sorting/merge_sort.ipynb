```python
# divide and merge

nums = [3,5,6,4,8,9,10,7,1] 

def merge_sort(arr):
    if len(arr)<=1:
        return(arr)
    
    mid = len(arr)//2
    left_arr=nums[:mid]
    right_arr=nums[mid:]
    left = merge_sort(right_arr)
    right = merge_sort(left_arr)
    
    return merge_array(left,right)
    
```
